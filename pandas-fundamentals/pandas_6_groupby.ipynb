# .groupby methodu

```python
# .groupby(stün ismi) bu kullanım ile bir sütundaki değerleri göre gruplama yapılabilir.
# devamında aggregate function'ına bakmalıyız.
```

- mean(): mean of groups
- sum(): sum of  groups
- size(): group sizes
- count(): count of group
- std(): standart variance of groups
- var(): variance of groups
- sem(): standart error of the mean of groups
- describe(): generates descirptive statistics
- first(): first of group values
- last(): last of group values
- nth(): take mth value , or a subset if in is a list
- min(): min of group values
- max(): max of group values
### Full list online documentation:https://pandas.pydata.org/docs/reference/groupby.html

```python
#soru mpg performansını yıllara göre kıyasla
```

```python
df.groupby('model_year')['mpg'].mean()
```
