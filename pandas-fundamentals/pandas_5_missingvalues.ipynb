# Missing valuelarla yapılabilecek 3 şey vardır.

```python
# 1- Dokunmadan bırakabiliriz.
# 2- Drop edebiliriz.(satır veya sütun bazlı)
# 3- Mantıklı bir şekilde doldurabiliriz.
```

```python
df
```

```python
# 1. yol: yapılacak hiçbir şey yok.
```

```python
# 2. yol:
```

```python
df.dropna()
```

```python
df.dropna(how = 'all')
#hepsi nan olanları düşürür.
```

```python
df.dropna(thresh = 1)
# bir tane bile nan olamyan veri varsa o satırı tut.
```

```python
# 3. yol: Fill the data (datayı dolurmak)
```

```python
df.head()
```

```python
df.dropna(how = 'all',inplace=True)
```

```python
df['pre_movie_score'].fillna(9.6)
# manuel doldurma
```

```python
df['pre_movie_score'].fillna(df['pre_movie_score'].mean())
#sütunda bulunan değerlerin ortalaması ile doldurma.
```
