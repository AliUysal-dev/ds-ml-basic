### 1. yol

```python
# def fiyat_kategorisi(x):
#  if x < 10 :
#    return '$'
#  elif 10 <= x < 30 :
#    return '$$'
#  else :
#    return '$$$'
```

### 2. yol

```python
df['fiyat_kategorisi'] = df['total_bill'].apply(lambda x : '$' if x < 10 else '$$' if 10<= x <30 else '$$$' )
```

```python
df.sample(2)
```

### 3. yol (en hızlısı)

```python
df.drop('fiyat_kategorisi', axis = 1 , inplace = True)
```

```python
def fiyat_kategorisi_vectorized(bill):
  categories = np.full_like(bill, '$', dtype = 'U3')

  categories[(bill > 10) & (bill < 30)] = '$$'

  categories[(bill >= 30 )] = '$$$'

  return categories
```

```python
df['fiyat_kategorisi'] = fiyat_kategorisi_vectorized(df['total_bill'].values)
```

```python
df.sample(3)
```

```python
df = df.drop('sex', axis=1)
display(df.head())
```
