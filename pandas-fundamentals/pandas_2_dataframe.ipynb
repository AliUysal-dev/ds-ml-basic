# DataFrame'de Sorgu İşlemleri

```python
df=pd.DataFrame({'year':[1776,1867,1821],'pop':[328,38,126],'GDP':[20.5,1.7,1.22]} , index=['USA','CANADA','MEXICO'] )
df
```

```python
#hangi ülkelerin 50 milyondan daha fazla popülasyonu var ?
```

```python
df[df['pop'] > 50]
```

```python
df[df['year']>1800]
```

```python
df = pd.read_csv('tips.csv')
df.head()
```

```python
df[df['total_bill']>40]
```

```python
df[df['sex']=='Male']
```

```python
df[(df['sex']=='Female') & (df['tip']>5)]
# & sembolü and yerine kullanırız ve parantezde göstermek zorundayız
```

```python
df[(df['day']=='Sun') & (df['total_bill']>30)]
```

```python
#veya | sembolü ile yapılır
```

```python
df[(df['day'] ==  'Sat') | (df['day'] == 'Sun')]
```

```python
df[df['day'].isin(['Sat','Sun','Thur'])]
# 3 tane veyalı sorgu yerine .isin methodu kullanılır
# içinde var mı demek.
```
