# Series oluşturma

```python
my_data = [1923,1776,1876]
my_idx = ['Türkiye','USA','Canada']
```

# Serilerde İşlemler

```python
q1 = {'Japan': 80, 'China': 450, 'India': 200, 'USA': 250}
q2 = {'Japan': 100,'China': 500, 'India': 210,'USA': 260}
```

```python
sales_q1 = pd.Series(q1)
sales_q2 = pd.Series(q2)
```

```python
sales_total = sales_q1 + sales_q2
```

```python
sales_total
```

```python
sales_total.index #index attribute'ı ile bakabiliriz.
```

```python
sales_total.keys() #keys() metoduyla da indexlerine bakabiliriz.
```

```python
sales_total.values #.values attribute'ı ile de değerlerine bakabiliriz.
```

```python
sales_total
```

```python
sales_total * 2
```

```python
sales_total / 10
```

```python
sales_q2 - sales_q1
```

```python
league_2022 = {
    'Fenerbahce': 51,
    'Galatasaray': 60,
    'Beşiktaş': 50,
    'Trabzonspor': 48
}
league_2021 = {
    'Fenerbahce': 73,
    'Galatasaray': 68,
    'Beşiktaş': 59,
    'Trabzonspor': 81
}
league_2020 = {
    'Fenerbahce': 82,
    'Galatasaray': 84,
    'Beşiktaş': 84,
    'Trabzonspor': 71
}
```

```python
lg_1 = pd.Series(league_2022)
lg_2 = pd.Series(league_2021)
lg_3 = pd.Series(league_2020)
```

```python
lg_1 + lg_2 + lg_3
```
