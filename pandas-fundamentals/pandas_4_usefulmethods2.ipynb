```python
df = pd.read_csv('tips.csv')
```

```python
df[df['total_bill'].between(10,20)]
#default ayarlarda 10 ve 20 dahildir
#inculusive paramtersiyle left ve right ile 10 veya 20 'nin dahil olup olmayacağını seçebiliriz.
```
