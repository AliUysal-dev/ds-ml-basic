```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
yaslistesi = [10,20,30,30,30,40,50,60,70,75]
kilolistesi = [20,60,80,85,86,87,70,90,95,90]
```

```python
numpyyaslistesi = np.array(yaslistesi)
numpykilolistesi = np.array(kilolistesi)
```

```python
numpykilolistesi
```

```python
plt.plot(numpyyaslistesi,numpykilolistesi,'r')
# r-red ifade eder ve grafiğimiz kırmızı oldu.
```

```python
plt.plot(numpykilolistesi,numpyyaslistesi,'y')
#x ve y eksenini değiştirebiliyoruz.
```

```python
plt.plot(numpyyaslistesi,numpykilolistesi,'y')
plt.xlabel('yas')
plt.ylabel('kilo')
plt.title("kilo'nun yaşa göre değişim grafiği")
 # burada ise x eksini ve y eksenini adlandırdık.
# tabi aynı zamanda title fonksiyou ile başlıkda koyabiliriz. 
```

```python
npd1 = np.linspace(0,10,20)
npd1
```

```python
npd2 = npd1 **3
npd2
```

```python
plt.plot(npd1,npd2,'r*-')
# r-- kesikli gösterim
# r*- yıldızlı gösterim
```

```python
plt.subplot(1,2,1)
plt.plot(npd1,npd2,'r')
plt.subplot(1,2,2)
plt.plot(npd2,npd1,'g')
#farklı grafiklerin birlikte gösterimi.
```

```python
figure1 = plt.figure()
# görselleştirmenin boş halinin ismi.
```

```python
figure1 = plt.figure()
figureaxes = figure1.add_axes([0.1,0.1,0.5,0.5])
#eksenine göre artış değerini ve boyutunu yazabiliyoruz.
figureaxes.plot(npd1,npd2,'g')
figureaxes.set_xlabel("x ekseni veri ismi")
figureaxes.set_ylabel("y ekseni veri ismi")
figureaxes.set_title('Başlık')
```

```python
figure2 = plt.figure()
eksen1 = figure2.add_axes([0.1,0.1,0.5,0.5])
eksen2 = figure2.add_axes([0.2,0.1,0.3,0.3])

eksen1.plot(npd1,npd2,'g')
eksen1.set_xlabel('x ekseni')
eksen1.set_ylabel('y ekseni')
eksen1.set_title('başlık')

eksen2.plot(npd2,npd1,'b')
eksen2.set_xlabel('x ekseni')
eksen2.set_label('y ekseni')
eksen2.set_title('başlık')
```

```python
(benimF,benimE) = plt.subplots(nrows=1,ncols=2)
type(benimE)

for eksen in benimE:
    eksen.plot(npd1,npd2,'r')
    eksen.set_xlabel('x ekseni')
plt.tight_layout()
```

```python
newf = plt.figure(dpi=75)
#aynı zamanda burada örneğin figsize = 3,3 vererek de değiştirebiliriz.
# dpi parametresi kalitesini değiştirebiliyoruz.
yenie = newf.add_axes([0.1,0.1,0.6,0.6])
yenie.plot(npd1,npd1**2,label = 'npd**2')
# ne verisi olduğunu belirttik npd**2
yenie.plot(npd1,npd1**3, label = 'npd**3')
yenie.legend(loc=2)
#burada örneğin loc parametresi kullanrak yerini değiştirebiliyoruz.
#sacefig parametresi ile kayıt edilebilir .png dosyası olarak.
```
