# çok boyutlu arraylerde iteration(yineleme)

```python
ars1d = np.arange(10)
ars1d
```

```python
for i in ars1d:
    print(i)
```

```python
ars2d = np.arange(20).reshape(4,5)
ars2d
```

```python
for x in ars2d:
  print(x)
```

```python
for x in ars2d:
    for y in x:
        print(y)
```

```python
ars3d = np.arange(40).reshape(2,2,10)
ars3d
```

```python
for x in ars3d:
    for y in x:
        for z in y:
            print(z)
```

```python
#for kullanımı hiçbir şekilde önerilmiyor
```

#numpy arraylerini birleştirme (concatenate)

```python
#numpy arraylerini birleştirme işlemi .concatenate() fonksiyonu ile yapılır
```

```python
#np.concatenate([ar1,ar2])
```

```python
ar1 = np.array([1,2,3,4])
ar2 = np.array([5,6,7,8])
```

```python
np.concatenate([ar1,ar2])
```

```python
np.concatenate([ar1,ar2],axis=0)
# yan yana ekleme
```

```python
np.concatenate([ar1,ar2],axis=1)
# stün halinde ekleme
```

```python
# bir boyutlu arraylerde olmuyor.
```

```python
ars1 = np.array([[1,2,3],[4,5,6]])
ars2 = np.array([[7,8,9],[10,11,12]])
```

```python
np.concatenate([ars1,ars2],axis=1)
```

```python
np.concatenate([ars1,ars2],axis=0)
```
